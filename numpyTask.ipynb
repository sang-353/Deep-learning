```python
import numpy as np
```

```python
arr = np.array([28,30,29,31,32,30,29])
print(arr.mean())
print(arr.max())
print(arr.min())
print(len(arr[arr > 30]))
print(np.where(arr > 30, arr, 0))
print(np.count_nonzero(np.where(arr > 30, arr, 0)))
```

```python
marks = np.array([85,90,78,92,88])
print(np.mean(marks))
print(np.sort(marks))
print(np.median(marks))
print(np.std(marks))
print(marks/10)
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print(A+B)
print(A * B)
print(A @ B)
```

```python
arr2 = np.random.randint(0,10,(3,4))
print(arr2)
print(np.max(arr2,axis = 0))
print(np.min(arr2,axis = 1))
print(np.where(arr2 % 2 == 0, arr2, -1))
arr2[arr2 % 2 == 1] = -1
print(arr2)
```

```python
arr3 = np.arange(1,13)
print(arr3)
print(np.reshape(arr3,(3,4)))
arr3 = np.reshape(arr3,(3,4))
print(np.sum(arr3,axis=1))
print(np.mean(arr3,axis=0))
print(np.reshape(arr3,(12)))
```

```python
arr4 = np.random.randint(0,20,(5,5))
print(arr4)
print(arr4[arr4 > 10])
print(np.where(arr4 > 10, 0, arr4))
```

```python
arrA = np.array([1,2,3])
arrB = np.array([4,5,6])
print(np.concatenate((arrA, arrB)))
```

```python
arr5 = np.array([2,1,2,3,1,4,3])
unique_var,counts = np.unique(arr5,return_counts=True)
print(unique_var)
print(counts)
for val in unique_var:
    print(len(arr5[arr5 == val]))
```
