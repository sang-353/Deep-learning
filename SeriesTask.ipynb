```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
# 1
scores = pd.Series(np.random.randint(50,101, 10),index=['Student'+str(i) for i in range(1,11)], name='Score')
print(scores)
print('Mean:', scores.mean())
means = scores.mean()
print('Max:', scores.max())
print('Min:', scores.min())
scores[scores > means]
# print('Number of students scoring above mean:', scores[scores > means],len(scores[scores > means]))
```

```python
# 2
temperatures = pd.Series([28,31,29,32,30,27,33],index=['Mon','Tue','Wed','Thu','Fri','Sat','Sun'], name='Temperature')
print(temperatures)
len(temperatures[temperatures>30])
print(temperatures.mean())
print(temperatures.sort_values(ascending=False))
diffs = temperatures.diff().abs()
diffs.sort_values(ascending=False).keys()[:2].to_list()
print(diffs)
```

```python
# 3
prices = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2026-09-01', periods=10), name='Price')
a = prices.pct_change()
print(a.idxmax())
print(a.idxmin())
print(a.std())
```
