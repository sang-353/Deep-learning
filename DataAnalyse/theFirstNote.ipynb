```python
arr1 = np.zeros((2,3))
print(arr1)
```

```python
arr2 = np.ones((2,3),dtype=int)
print(arr2)
```

```python
arr3 = np.empty((4,3),dtype=int)
print(arr3)
```

```python
arr4 = np.full((2,3),45)
print(arr4)
```

```python
lists = [[1,2,3],[4,5,6]]
arr5 = np.array(lists)
print(arr5)
```

```python
arr6 = np.full_like(arr5, 7)
print(arr6)
```

```python
arr7= np.arange(2,10,2)
print(arr7)
```

```python
arr8 = np.linspace(1,10,4,dtype=float)
print(arr8)
```

```python
arr9 = np.logspace((1,4),2,2,dtype=int)
print(arr9)
```

```python
arr10 = np.eye(3,3)
print(arr10)
```

```python
arr11 = np.diag([1,2,3])
print(arr11)
```

```python
np.random.seed(40)
arr12 = np.random.rand(2,3)
print(arr12)
arr13 = np.random.randn(3,4)
print(arr13)
arr14 = np.random.uniform(2,3, (2,3))
print(arr14)
```
