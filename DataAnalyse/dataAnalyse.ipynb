```python
# 企鹅数据分析
# 1. 导入必要库
import pandas as pd
# 2. 导入数据
df = pd.read_csv('../data/penguins.csv')
df.head()
df.info()
```

```python
# 3. 数据清洗
# 缺失值检查
print(df.isna().sum())
df.dropna(inplace=True)
```

```python
# 4. 数据特征的构造
df['sex'] = df['sex'].astype('category')
df['bill_ratio'] = df['bill_length_mm'] / df['bill_depth_mm']
```

```python
# 5. 数据分析
# 数据分箱-把体重分为三个等级
df['mass_level'] = pd.cut(df['body_mass_g'], bins=3, labels=['light', 'medium', 'heavy'])
print(df['mass_level'].value_counts())
# 按岛屿、性别分组分析
df.groupby(['sex', 'island']).agg({
    'body_mass_g': ['mean', 'count'],
    'bill_length_mm': ['mean', 'count'],
    'bill_depth_mm': ['mean', 'count'],
    'bill_ratio': ['mean', 'count']
})
```
